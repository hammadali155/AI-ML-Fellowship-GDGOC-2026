## Task 1: Handling Missing Data – Titanic Dataset

### Overview
Identify and handle missing values in the Titanic dataset.

### Technique Used
**Mean/Median Imputation** for continuous variables and **Mode Imputation** for categorical variables, along with **Dropping** high-null columns.

### Why?
- **Mean/Median Imputation:** Efficient for numerical data with small amounts of missing values. Median is preferred if outliers are present.
- **Mode Imputation:** Standard for categorical data.
- **Dropping:** Used for columns like 'Cabin' (or 'deck') where >70% of data is missing, making imputation unreliable.

```python
import pandas as pd
import numpy as np

# Load dataset
df_titanic = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv')

print("Missing values before preprocessing:")
print(df_titanic.isnull().sum())

# 1. Median Imputation for 'age'
df_titanic['age'] = df_titanic['age'].fillna(df_titanic['age'].median())

# 2. Mode Imputation for 'embarked' and 'embark_town'
df_titanic['embarked'] = df_titanic['embarked'].fillna(df_titanic['embarked'].mode()[0])
df_titanic['embark_town'] = df_titanic['embark_town'].fillna(df_titanic['embark_town'].mode()[0])

# 3. Dropping 'deck' (Cabin equivalent) due to excessive missing data
if 'deck' in df_titanic.columns:
    df_titanic.drop(columns=['deck'], inplace=True)

print("\nMissing values after preprocessing:")
print(df_titanic.isnull().sum())
```

### Results & Observations
- All missing values in 'age', 'embarked', and 'embark_town' were successfully filled.
- Dropping the 'deck' column cleaned the dataset significantly without losing critical predictive power as the column was mostly empty.
