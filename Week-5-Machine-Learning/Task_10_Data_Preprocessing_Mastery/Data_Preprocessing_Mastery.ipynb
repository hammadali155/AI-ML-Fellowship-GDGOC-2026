## Task 1: Handling Missing Data – Titanic Dataset

### Overview
Identify and handle missing values in the Titanic dataset.

### Technique Used
**Mean/Median Imputation** for continuous variables and **Mode Imputation** for categorical variables, along with **Dropping** high-null columns.

### Why?
- **Mean/Median Imputation:** Efficient for numerical data with small amounts of missing values. Median is preferred if outliers are present.
- **Mode Imputation:** Standard for categorical data.
- **Dropping:** Used for columns like 'Cabin' (or 'deck') where >70% of data is missing, making imputation unreliable.

```python
import pandas as pd
import numpy as np

# Load dataset
df_titanic = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv')

print("Missing values before preprocessing:")
print(df_titanic.isnull().sum())

# 1. Median Imputation for 'age'
df_titanic['age'] = df_titanic['age'].fillna(df_titanic['age'].median())

# 2. Mode Imputation for 'embarked' and 'embark_town'
df_titanic['embarked'] = df_titanic['embarked'].fillna(df_titanic['embarked'].mode()[0])
df_titanic['embark_town'] = df_titanic['embark_town'].fillna(df_titanic['embark_town'].mode()[0])

# 3. Dropping 'deck' (Cabin equivalent) due to excessive missing data
if 'deck' in df_titanic.columns:
    df_titanic.drop(columns=['deck'], inplace=True)

print("\nMissing values after preprocessing:")
print(df_titanic.isnull().sum())
```

### Results & Observations
- All missing values in 'age', 'embarked', and 'embark_town' were successfully filled.
- Dropping the 'deck' column cleaned the dataset significantly without losing critical predictive power as the column was mostly empty.

## Task 2: Feature Encoding – Car Evaluation Dataset

### Overview
Encode categorical variables in the Car Evaluation dataset.

### Technique Used
- **One-Hot Encoding (OHE):** For nominal categorical variables.
- **Label Encoding:** For ordinal categorical variables (or comparison).

### Why?
- **One-Hot Encoding:** Prevents the model from assuming an mathematical order between categories that don't have one (e.g., color).
- **Label Encoding:** Useful when there is a clear order (e.g., small, medium, large) despite being categorical.

```python
from sklearn.preprocessing import LabelEncoder

# Load Car Evaluation dataset
url_car = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"
columns_car = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'class']
df_car = pd.read_csv(url_car, names=columns_car)

print("Original Data Sample:")
print(df_car.head())

# 1. Label Encoding
le = LabelEncoder()
df_label = df_car.copy()
for col in df_label.columns:
    df_label[col] = le.fit_transform(df_label[col])

# 2. One-Hot Encoding
df_ohe = pd.get_dummies(df_car, columns=['lug_boot', 'safety'])

print("\nShape after Label Encoding:", df_label.shape)
print("Shape after One-Hot Encoding (on 2 columns):", df_ohe.shape)
print("\nOne-Hot Encoded columns sample:")
print(df_ohe.columns[-10:])
```

### Results & Observations
- **Label Encoding** maintained the original column count, merely mapping strings to integers.
- **One-Hot Encoding** increased the feature space (dimensionality) by creating binary columns for each category.
- **Observation:** OHE is better for non-ordinal data, while Label Encoding is more memory-efficient but might introduce artificial ordering.

## Task 3: Feature Scaling – Wine Quality Dataset

### Overview
Apply normalization and standardization to features in the Wine Quality dataset.

### Technique Used
- **Normalization (Min-Max Scaling):** Scales data to a fixed range [0, 1].
- **Standardization (Z-score Scaling):** Scales data to have a mean of 0 and standard deviation of 1.

### Why?
- **Normalization:** Useful when features have different ranges but no extreme outliers, and when the algorithm requires bounded inputs (e.g., Neural Networks).
- **Standardization:** More robust to outliers and preferred for algorithms that assume a Gaussian distribution of features (e.g., Linear Regression, SVM).

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Load Wine Quality dataset (Red Wine)
url_wine = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df_wine = pd.read_csv(url_wine, sep=';')

# Selecting a subset of features for demonstration
features_to_scale = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar']
data_subset = df_wine[features_to_scale]

# 1. Normalization
scaler_norm = MinMaxScaler()
df_normalized = pd.DataFrame(scaler_norm.fit_transform(data_subset), columns=features_to_scale)

# 2. Standardization
scaler_std = StandardScaler()
df_standardized = pd.DataFrame(scaler_std.fit_transform(data_subset), columns=features_to_scale)

print("Original Data Statistics (Fixed Acidity):")
print(f"Mean: {data_subset['fixed acidity'].mean():.2f}, Std: {data_subset['fixed acidity'].std():.2f}, Min: {data_subset['fixed acidity'].min()}, Max: {data_subset['fixed acidity'].max()}")

print("\nNormalized Data Statistics (Fixed Acidity):")
print(f"Mean: {df_normalized['fixed acidity'].mean():.2f}, Std: {df_normalized['fixed acidity'].std():.2f}, Min: {df_normalized['fixed acidity'].min()}, Max: {df_normalized['fixed acidity'].max()}")

print("\nStandardized Data Statistics (Fixed Acidity):")
print(f"Mean: {df_standardized['fixed acidity'].mean():.2f}, Std: {df_standardized['fixed acidity'].std():.2f}, Min: {df_standardized['fixed acidity'].min()}, Max: {df_standardized['fixed acidity'].max()}")
```

### Results & Observations
- **Normalization** successfully squeezed all values into the [0, 1] range.
- **Standardization** centered the data around 0 with a unit standard deviation.
- **Observation:** Scaling is crucial for distance-based algorithms to prevent features with larger scales from dominating the model.

## Task 4: Handling Outliers – Boston Housing Dataset

### Overview
Detect and treat outliers in the Boston Housing dataset.

### Technique Used
- **Z-score Method:** Identifies outliers as data points that are more than 3 standard deviations from the mean.
- **Interquartile Range (IQR) Method:** Identifies outliers as points outside [Q1 - 1.5*IQR, Q3 + 1.5*IQR].

### Why?
- **Z-score:** Standard statistical approach for normally distributed data.
- **IQR:** Robust to the underlying distribution and effective for skewed data, which is common in housing price features.

```python
from scipy import stats

# Load Boston Housing dataset (Raw link from GitHub)
url_boston = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"
df_boston = pd.read_csv(url_boston)

print("Original Dataset Shape:", df_boston.shape)

# Feature to check: 'rm' (average number of rooms per dwelling)
feature = 'rm'

# 1. Z-score Approach
z_scores = np.abs(stats.zscore(df_boston[feature]))
df_z_clean = df_boston[z_scores < 3]

# 2. IQR Approach
Q1 = df_boston[feature].quantile(0.25)
Q3 = df_boston[feature].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_iqr_clean = df_boston[(df_boston[feature] >= lower_bound) & (df_boston[feature] <= upper_bound)]

print(f"\nOutliers detected in '{feature}' using Z-score ( > 3 std): {len(df_boston) - len(df_z_clean)}")
print(f"Outliers detected in '{feature}' using IQR: {len(df_boston) - len(df_iqr_clean)}")

print("\nCleaned Dataset Shape (IQR method):", df_iqr_clean.shape)
```

### Results & Observations
- **IQR method** identified more outliers compared to the Z-score (3 threshold) for the 'rm' feature.
- **Observation:** Treating outliers prevents models like Linear Regression from assigning excessive weight to extreme, potentially erroneous, data points.

## Task 5: Data Imputation (Advanced) – Retail Sales Dataset

### Overview
Handle missing values using advanced techniques: KNN Imputation and MICE (Iterative Imputer).

### Technique Used
- **KNN Imputer:** Fills missing values based on the 'k' most similar neighbors in the feature space.
- **MICE (Iterative Imputer):** Models each feature with missing values as a function of others in a round-robin fashion.

### Why?
- **KNN:** Captures local patterns and correlations between features, often more accurate than simple mean/median for non-linear data.
- **MICE:** Account for the statistical relationship between features, maintaining the distribution and correlations better than univariate methods.

```python
from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer


# Using a subset of the Retail Sales dataset (synthetic missingness for demo)
url_retail = "https://raw.githubusercontent.com/vkrit/data-science-class/master/WA_Fn-UseC_-Sales-Win-Loss.csv"
df_retail = pd.read_csv(url_retail).select_dtypes(include=[np.number]).iloc[:500, :5] # Numeric subset

# Introduce synthetic missing values (10%)
np.random.seed(42)
df_missing = df_retail.mask(np.random.random(df_retail.shape) < 0.1)

print("Missing values per column before imputation:")
print(df_missing.isnull().sum())

# 1. KNN Imputation
knn_imputer = KNNImputer(n_neighbors=5)
df_knn = pd.DataFrame(knn_imputer.fit_transform(df_missing), columns=df_missing.columns)

# 2. MICE (Iterative Imputer)
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice = pd.DataFrame(mice_imputer.fit_transform(df_missing), columns=df_missing.columns)

print("\nMissing values after KNN:", df_knn.isnull().sum().sum())
print("Missing values after MICE:", df_mice.isnull().sum().sum())

print("\nComparison of a specific imputed value (Row 0):")
print(f"Original: {df_retail.iloc[0, 0]}")
print(f"KNN Imputed: {df_knn.iloc[0, 0]:.2f}")
print(f"MICE Imputed: {df_mice.iloc[0, 0]:.2f}")
```

### Results & Observations
- Both **KNN** and **MICE** successfully imputed all missing values.
- **Observation:** Advanced methods are superior for multi-variate datasets as they leverage relationships between columns rather than treating them in isolation.

## Task 6: Feature Engineering – Heart Disease Dataset

### Overview
Create new derived features from existing ones in the Heart Disease dataset.

### Technique Used
- **Binning (Discretization):** Converting continuous age into discrete age groups.
- **Arithmetic Composition:** Combining features like 'chol' (cholesterol) and 'trestbps' (resting blood pressure) to create a 'risk_score'.
- **Boolean Transformation:** Creating flags for high-risk conditions.

### Why?
- **Binning:** Captures non-linear relationships where risk might jump at specific age thresholds rather than increasing linearly.
- **Derived Features:** Help the model learn interactions between variables (like BP and Cholesterol) that are medically significant.

```python
# Load Heart Disease dataset
url_heart = "https://raw.githubusercontent.com/kb22/Heart-Disease-Prediction/master/dataset.csv"
df_heart = pd.read_csv(url_heart)

# 1. Age Grouping
bins = [0, 30, 45, 60, 100]
labels = ['Young', 'Middle-Aged', 'Senior', 'Elderly']
df_heart['age_group'] = pd.cut(df_heart['age'], bins=bins, labels=labels)

# 2. Cholesterol to BP Ratio (Risk Factor)
df_heart['chol_bp_ratio'] = df_heart['chol'] / df_heart['trestbps']

# 3. High Risk Flag (Example: High Chol and High Age)
df_heart['high_risk_flag'] = ((df_heart['chol'] > 240) & (df_heart['age'] > 50)).astype(int)

print("Original Features (sample):")
print(df_heart[['age', 'chol', 'trestbps']].head())

print("\nEngineered Features (sample):")
print(df_heart[['age_group', 'chol_bp_ratio', 'high_risk_flag']].head())
```

### Results & Observations
- New features like **age_group** and **chol_bp_ratio** provide representational diversity.
- **Observation:** Engineered features often capture domain logic that raw data points alone might miss, potentially improving model interpretability and accuracy.

## Task 7: Variable Transformation – Bike Sharing Dataset

### Overview
Transform skewed variables in the Bike Sharing dataset using various mathematical techniques.

### Technique Used
- **Log Transformation:** Reduces right-skewness by compressing larger values.
- **Square Root Transformation:** Middling transformation for moderate skewness.
- **Box-Cox Transformation:** A parametric power transformation that finds the optimal power to stabilize variance and normalize data.

### Why?
- **Skewness Correction:** Many ML algorithms (like Linear Regression) perform better when features follow a normal distribution.
- **Variance Stabilization:** Transformations help in making the variance constant across the range of data.

```python
from scipy import stats

# Load Bike Sharing dataset
url_bike = "https://raw.githubusercontent.com/danwild/bike-share-prediction/master/Bike-Sharing-Dataset/day.csv"
df_bike = pd.read_csv(url_bike)

# Feature to transform: 'casual' (count of casual users) - usually highly skewed
feature_to_transform = 'casual'

# 1. Log Transformation (adding 1 to handle zeros)
df_bike['casual_log'] = np.log1p(df_bike[feature_to_transform])

# 2. Square Root Transformation
df_bike['casual_sqrt'] = np.sqrt(df_bike[feature_to_transform])

# 3. Box-Cox Transformation (only for positive data)
# Ensuring positive data by adding a small constant if needed
df_bike['casual_boxcox'], _ = stats.boxcox(df_bike[feature_to_transform] + 1)

print(f"Skewness of original '{feature_to_transform}': {df_bike[feature_to_transform].skew():.2f}")
print(f"Skewness after Log: {df_bike['casual_log'].skew():.2f}")
print(f"Skewness after Sqrt: {df_bike['casual_sqrt'].skew():.2f}")
print(f"Skewness after Box-Cox: {df_bike['casual_boxcox'].skew():.2f}")
```

### Results & Observations
- **Box-Cox** and **Log** transformations significantly reduced the skewness of the 'casual' user count.
- **Observation:** Reducing skewness makes the distribution more symmetric, allowing models to better capture the underlying patterns without being biased by extreme tails.

## Task 8: Feature Selection – Diabetes Dataset

### Overview
Select the most important features in the Pima Indians Diabetes dataset using different techniques.

### Technique Used
- **Correlation Analysis:** Identifies linear relationships between features and the target.
- **Mutual Information:** Measures the dependency between variables (captures non-linear relationships).
- **Recursive Feature Elimination (RFE):** Iteratively removes features and builds a model to identify the most predictive ones.

### Why?
- **Dimensionality Reduction:** Reducing the number of features simplifies the model and reduces overfitting.
- **Improved Performance:** Removing noise or redundant features often leads to better model accuracy and faster training.

```python
from sklearn.feature_selection import mutual_info_classif, RFE
from sklearn.linear_model import LogisticRegression

# Load Pima Indians Diabetes dataset
url_diabetes = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
columns_diabetes = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']
df_diabetes = pd.read_csv(url_diabetes, names=columns_diabetes)

X = df_diabetes.drop('Outcome', axis=1)
y = df_diabetes['Outcome']

# 1. Correlation Analysis
corr_matrix = df_diabetes.corr()
top_corr_features = corr_matrix['Outcome'].sort_values(ascending=False).index[1:4].tolist()

# 2. Mutual Information
mi_scores = mutual_info_classif(X, y, random_state=42)
mi_series = pd.Series(mi_scores, index=X.columns).sort_values(ascending=False)
top_mi_features = mi_series.index[:3].tolist()

# 3. Recursive Feature Elimination (RFE)
model = LogisticRegression(max_iter=1000)
rfe = RFE(estimator=model, n_features_to_select=3)
rfe.fit(X, y)
top_rfe_features = X.columns[rfe.support_].tolist()

print("Top 3 Features by Correlation:", top_corr_features)
print("Top 3 Features by Mutual Information:", top_mi_features)
print("Top 3 Features by RFE:", top_rfe_features)
```

### Results & Observations
- **Glucose** and **BMI** consistently appeared as top features across all selection methods.
- **Observation:** Different selection techniques might highlight different features based on whether they look for linear, non-linear, or model-specific importance. Using a consensus approach is often best.

## Task 9: Handling Imbalanced Data – Credit Card Fraud Detection

### Overview
Manage class imbalance in the Credit Card Fraud Detection dataset.

### Technique Used
- **Random Undersampling:** Reduces the number of samples in the majority class to match the minority class.
- **SMOTE (Synthetic Minority Over-sampling Technique):** Creates synthetic examples of the minority class by interpolating between existing ones.

### Why?
- **Bias Prevention:** ML models trained on imbalanced data often become biased towards the majority class, leading to poor performance on detecting the rare but critical minority class (fraud).
- **Evaluation Accuracy:** Balancing the classes allows for more meaningful evaluation metrics like Precision-Recall and F1-score.

```python
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from collections import Counter

# Load a small sample of Credit Card Fraud dataset
url_fraud = "https://raw.githubusercontent.com/nsethi31/Kaggle-Data-Credit-Card-Fraud-Detection/master/creditcard.csv"
df_fraud = pd.read_csv(url_fraud).sample(5000, random_state=42) # Sample for speed

X = df_fraud.drop('Class', axis=1)
y = df_fraud['Class']

print("Original class distribution:", Counter(y))

# 1. Random Undersampling
rus = RandomUnderSampler(random_state=42)
X_under, y_under = rus.fit_resample(X, y)
print("Distribution after Undersampling:", Counter(y_under))

# 2. SMOTE (Oversampling)
# Note: In real scenarios, apply SMOTE only on the training set
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)
print("Distribution after SMOTE:", Counter(y_smote))
```

### Results & Observations
- **Undersampling** balanced the classes by reducing the majority class, significantly shrinking the dataset.
- **SMOTE** balanced the classes by synthetically generating 4900+ fraud samples, maintaining the original dataset size but increasing the fraud representation.
- **Observation:** SMOTE is generally preferred as it preserves all majority class information, whereas undersampling discards it.

## Task 10: Combining Multiple Datasets – MovieLens Dataset

### Overview
Merge multiple related datasets from the MovieLens 100k collection (Ratings, Movies, and Users) into a single unified DataFrame.

### Technique Used
- **Inner Join:** Combining ratings with movie metadata using `movie_id`.
- **Inner Join:** Combining the resulting DataFrame with user demographics using `user_id`.

### Why?
- **Data Consolidation:** Real-world data is often siloed in different tables. Combining them is essential to perform comprehensive analysis, such as understanding which demographics prefer which genres.
- **Relational Integrity:** Using common keys (IDs) ensures that the correct information is mapped across datasets.

```python
# Load MovieLens 100k datasets
data_url = "https://raw.githubusercontent.com/Kostis-S-Z/moviestream_task/master/ml-100k/u.data"
items_url = "https://raw.githubusercontent.com/Kostis-S-Z/moviestream_task/master/ml-100k/u.item"
users_url = "https://raw.githubusercontent.com/Kostis-S-Z/moviestream_task/master/ml-100k/u.user"

# 1. Load Ratings
ratings_cols = ['user_id', 'movie_id', 'rating', 'timestamp']
df_ratings = pd.read_csv(data_url, sep='\t', names=ratings_cols)

# 2. Load Movies (Item information)
movie_cols = ['movie_id', 'movie_title', 'release_date', 'video_release_date', 'IMDb_URL', 'unknown', 'Action', 'Adventure', 'Animation', 'Childrens', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
df_movies = pd.read_csv(items_url, sep='|', names=movie_cols, encoding='latin-1')

# 3. Load Users
user_cols = ['user_id', 'age', 'gender', 'occupation', 'zip_code']
df_users = pd.read_csv(users_url, sep='|', names=user_cols)

# Step 1: Merge Ratings and Movies
df_merged = pd.merge(df_ratings, df_movies[['movie_id', 'movie_title']], on='movie_id')

# Step 2: Merge the result with User info
df_final = pd.merge(df_merged, df_users, on='user_id')

print("Shape of final merged dataset:", df_final.shape)
print("\nFirst few rows of the consolidated data:")
print(df_final[['user_id', 'movie_title', 'rating', 'age', 'gender']].head())
```

### Results & Observations
- Three separate datasets were successfully merged into a single DataFrame with 100,000 ratings.
- **Observation:** Merging allows for the enrichment of sparse data (like ratings) with descriptive attributes (movie titles and user demographics), enabling more complex feature engineering.

## Task 11: Dimensionality Reduction – MNIST Dataset

### Overview
Reduce the feature space of handwritten digits (MNIST-like) while preserving variance using PCA and t-SNE.

### Technique Used
- **PCA (Principal Component Analysis):** A linear dimensionality reduction technique that finds the principal components (orthogonal directions) that capture the maximum variance.
- **t-SNE (t-distributed Stochastic Neighbor Embedding):** A non-linear technique specifically designed for high-dimensional data visualization by preserving local structures.

### Why?
- **Speed & Efficiency:** Reducing the number of features significantly speeds up training of downstream models.
- **Visualization:** t-SNE is exceptionally good at clustering similar data points in 2D or 3D space, which is useful for data exploration and understanding class separation.

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.datasets import load_digits

# Load digits dataset (mini-MNIST)
digits = load_digits()
X, y = digits.data, digits.target

print(f"Original Feature Dimension: {X.shape[1]}")

# 1. PCA: Reduce to 10 components
pca = PCA(n_components=10)
X_pca = pca.fit_transform(X)
explained_variance = np.sum(pca.explained_variance_ratio_)

# 2. t-SNE: Reduce to 2 components for visualization
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne = tsne.fit_transform(X)

print(f"Dimensions after PCA: {X_pca.shape[1]}")
print(f"Cumulative Explained Variance (10 components): {explained_variance:.2f}")
print(f"Dimensions after t-SNE: {X_tsne.shape[1]}")
```

### Results & Observations
- **PCA** reduced the feature from 64 to 10 while retaining most of the important variance.
- **t-SNE** successfully mapped the 64-dimensional data into a 2D space.
- **Observation:** PCA is linear and focuses on global variance, whereas t-SNE is non-linear and focuses on local similarities between points.

## Task 12: Text Preprocessing – IMDB Movie Reviews Dataset

### Overview
Clean and preprocess raw text data from IMDB movie reviews to make it suitable for machine learning models.

### Technique Used
- **Lowercasing:** Converting all text to lowercase for consistency.
- **Tokenization:** Splitting text into individual words (tokens).
- **Stopword Removal:** Removing common words (e.g., 'the', 'is') that don't carry significant meaning.
- **Lemmatization:** Reducing words to their base or dictionary form (lemma).

### Why?
- **Normalization:** Ensures that 'Good', 'good', and 'GOOD' are treated as the same feature.
- **Noise Reduction:** Removing stopwords and punctuation reduces the dimensionality and focuses on sentiment-carrying words.
- **Improved Retrieval:** Lemmatization maps related word forms to the same root, helping the model generalize better.

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Ensure NLTK resources are available (in a real notebook, these would be downloaded once)
try:
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()
except:
    # Fallback if nltk data is not downloaded
    nltk.download('stopwords')
    nltk.download('wordnet')
    nltk.download('omw-1.4')
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    # 1. Lowercasing
    text = text.lower()
    # 2. Removing punctuation and special characters
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # 3. Tokenization
    tokens = text.split()
    # 4. Stopword Removal and Lemmatization
    cleaned_tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return " ".join(cleaned_tokens)

# Load a small sample of IMDB reviews
url_imdb = "https://raw.githubusercontent.com/Ankit152/IMDB-sentiment-analysis/master/IMDB-Dataset.csv"
df_imdb = pd.read_csv(url_imdb).head(5)

print("Original Review Sample:")
print(df_imdb['review'].iloc[0][:150] + "...")

# Apply Preprocessing
df_imdb['cleaned_review'] = df_imdb['review'].apply(preprocess_text)

print("\nPreprocessed Review Sample:")
print(df_imdb['cleaned_review'].iloc[0][:150] + "...")
```

### Results & Observations
- Raw reviews were successfully cleaned of HTML tags, punctuation, and stopwords.
- **Observation:** Text preprocessing significantly reduces the vocabulary size and transforms unstructured text into a more structured format for vectorization (like TF-IDF or Word Embeddings).

## Task 13: Time-Series Preprocessing – Air Quality Dataset

### Overview
Handle missing timestamps, perform resampling, and apply smoothing to the Air Quality time-series dataset.

### Technique Used
- **DateTime Casting:** Converting date/time strings into proper `datetime64` objects.
- **Resampling:** Changing the frequency of data points (e.g., from hourly to daily) to reduce noise.
- **Moving Average (Smoothing):** Calculating the rolling mean to identify longer-term trends.
- **Forward Fill (ffill):** Handling missing values by propagating the last valid observation forward.

### Why?
- **Temporal Consistency:** Machine learning models for time-series require a continuous and evenly spaced time index.
- **Noise Reduction:** Raw sensor data is often noisy; resampling and smoothing help in extracting the underlying signal.
- **Handling Gaps:** Real-world sensors often fail or experience gaps; imputation like forward fill maintains continuity without introducing future bias.

```python
# Load Air Quality dataset (sample with Date and Time)
url_air = "https://raw.githubusercontent.com/asharvi1/UCI-Air-Quality-Data/master/AirQualityUCI.csv"
df_air = pd.read_csv(url_air, sep=';', decimal=',').head(500)

# 1. Date and Time Preprocessing
df_air['datetime'] = pd.to_datetime(df_air['Date'] + ' ' + df_air['Time'], format='%d/%m/%Y %H.%M.%S', errors='coerce')
df_air = df_air.dropna(subset=['datetime']).set_index('datetime')

# 2. Resampling (Hourly to Daily Mean)
df_daily = df_air[['CO(GT)', 'T', 'RH']].resample('D').mean()

# 3. Handling missing values in resampled data
df_daily = df_daily.ffill()

# 4. Smoothing (3-Day Rolling Mean)
df_daily['CO_smoothed'] = df_daily['CO(GT)'].rolling(window=3).mean()

print("Original Data Sampling (first 3 rows):")
print(df_air[['CO(GT)', 'T']].head(3))

print("\nDaily Resampled & Smoothed Data (first 5 rows):")
print(df_daily.head(5))
```

### Results & Observations
- The hourly sensor data was successfully cleaned and aggregated into a daily format.
- **Observation:** Smoothing through rolling averages helps in visualizing the overall trend of air pollutants by removing short-term fluctuations.
